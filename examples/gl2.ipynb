```python
import poppy
import jax
```

```python
p = 2
n = 3
f = poppy.field(p,n)

print(f'GL2(F{p**n}) has order {len(poppy.gl2(f)[0])}.')
print(f'PGL2(F{p**n}) has order {len(poppy.pgl2(f)[0])}.')
print(f'SL2(F{p**n}) has order {len(poppy.sl2(f)[0])}.')
print(f'PSL2(F{p**n}) has order {len(poppy.psl2(f)[0])}.')
```

```python
p = 11
dtype = 'int64'
psl2, i = poppy.psl2mod(p, dtype = dtype)
PSL2 = jax.vmap(poppy.decode, in_axes = (0,None,None))(psl2,p,dtype).swapaxes(0,1)
TABLE = poppy.outer44(PSL2,PSL2,p)
TITLE = f'PSL2(Z/{p}) TIMES TABLE'

poppy.plot(poppy.unblock(TABLE), title = TITLE, size = 3)
poppy.plot(poppy.trace(TABLE,p), title = TITLE+' TRACE', size = 3)
```

```python
p = 11
dtype = 'int64'
pgl2, i = poppy.pgl2mod(p, dtype)
PGL2 = jax.vmap(poppy.decode, in_axes = (0,None,None))(pgl2,p, dtype).swapaxes(0,1)
TABLE = poppy.outer44(PGL2,PGL2,p)
TITLE = f'PGL2(Z/{p}) TIMES TABLE'

poppy.plot(poppy.unblock(TABLE), title = TITLE, size = 3)
poppy.plot(poppy.trace(TABLE,p), title = TITLE+' TRACE', size = 3)
```
