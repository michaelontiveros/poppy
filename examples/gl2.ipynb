```python
import poppy
import jax
```

```python
p = 2
n = 3

f = poppy.field(p,n)

print(f'GL2(F{p**n}) has order {len(poppy.gl2(f)[0])}.')
print(f'PGL2(F{p**n}) has order {len(poppy.pgl2(f)[0])}.')
print(f'SL2(F{p**n}) has order {len(poppy.sl2(f)[0])}.')
print(f'PSL2(F{p**n}) has order {len(poppy.psl2(f)[0])}.')
```

```python
p = 11

psl2, i = poppy.psl2mod(p)
PSL2 = jax.vmap(poppy.decode, in_axes = (0,None))(psl2,p).swapaxes(0,1)
TABLE = poppy.outer44(PSL2,PSL2,p)
TITLE = f'PSL2(Z/{p}) MULTIPLICATION TABLE'

poppy.plot(poppy.unblock(TABLE), title = TITLE, size = 12)
poppy.plot(poppy.trace(TABLE,p), title = TITLE+' TRACE', size = 12)
```
